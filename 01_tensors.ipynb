```python
import os
os.environ['TF_CPP_MIN_LOG_LEVEL'] = '2'  # Suppress TensorFlow logging (1)
import tensorflow as tf
```

# Rank 1 Tensor

```python
# Rank 1 Tensor (Vector)
x = tf.constant(4 , shape=(1,2) , name="X")
y = tf.constant(4)

print ("Value of Tensor X" , x)
print ("Value of Tensor Y" , y)

print(x + y)  # Addition
print(x - y)  # Subtraction
```

# Rank 2 Tensor

```python
# Rank 2 Tensor (Matrix)
a = tf.constant([[1,2],[3,4]] , name="A")
b = tf.constant([[5,6],[7,8]] , name="B")
print ("Value of Tensor A" , a)
print ("Value of Tensor B" , b)

print(a + b)  # Addition
print(a - b)  # Subtraction
```

```python
z = tf.zeros((2,3) , name="Z")  # Rank 2 Tensor of zeros
print ("Value of Tensor Z \n" , z)
print (end="\n")

x = tf.ones((2,3) , name="X")  # Rank 2 Tensor of ones
print ("Value of Tensor X \n" , x)
print (end="\n")
# Diagonal matrix
y = tf.eye(3 , name="Y")  # Rank 2 Identity Tensor
print ("Value of Tensor Y \n" , y)
print (end="\n")
# Random
v = tf.random.normal((3,3) , mean=0 , stddev=1 , name="V")  # Rank 2 Tensor of random values from a normal distribution
print ("Value of Tensor V \n" , v)
print (end="\n")
# Uniform distribution
w = tf.random.uniform((3,3) , minval=0 , maxval=1 , name="W")  # Rank 2 Tensor of random values from a uniform distribution
print ("Value of Tensor W \n" , w)
print (end="\n")
```

# Rank 3 Tensor

```python
# Rank 3 Tensor
c = tf.constant([[[1,2 , 3],[3,4 , 5]],[[5,6, 7],[7,8, 9]]] , name="C")
print ("Value of Tensor C" , c)
print (end="\n")

d = tf.zeros((3,3,3) , name="D")  # Rank 3 Tensor of zeros
print ("Value of Tensor D" , d)
print (end="\n")
```

# Elementwise Operations

```python
a = tf.constant([1,2,3] , name="A")
b = tf.constant([4,5,6] , name="B")

res = tf.add(a,b)  # Element-wise addition
print ("Result of A + B \n" , res)

add_res = a + b
print ("Result of A + B using operator \n" , add_res)
print (end="\n")

#=======================================================

sub = tf.subtract(a,b)  # Element-wise subtraction
print ("Result of A - B \n" , sub)

sub_res = a - b
print ("Result of A - B using operator \n" , sub_res)
print (end="\n")

#=======================================================
# here we should put into consideration matrices multiplication rules of the shape of both
mul = tf.multiply(a,b)  # Element-wise multiplication
print ("Result of A * B \n" , mul)

mul_res = a * b
print ("Result of A * B using operator \n" , mul_res)
print (end="\n")

#=======================================================

div = tf.divide(a,b)  # Element-wise division
print ("Result of A / B \n" , div)

div_res = a / b
print ("Result of A / B using operator \n" , div_res)
print (end="\n")
```

# Slicing , Indexing

```python
x = tf.constant([[1,2 , 5 , 6],[3,4 , 6 , 7]] , name="X")
print (x[0])
print (x[1])
print (x[0][1])
print (x[:,0]) # all rows and the first column
print (x[:,1]) # all rows and the second column
print (x[0,:]) # all columns and the first row
```

# Reshaping

```python
x = tf.constant([[1,2],[3,4]] , name="X")
print (x)
x = tf.reshape(x , (4,1))  # Reshape to a column vector
print (x)
x = tf.reshape(x , (1,4))  # Reshape back to original shape
print (x)
# must have the same number of elements before and after reshaping
```

# Tensor Conversion


```python
import numpy as np
a = np.array([[1,2],[3,4]])
print (a)

x = tf.convert_to_tensor(a , dtype=tf.int32)  # Convert NumPy array to TensorFlow tensor
print (x)
```
